```python
# install PuLP if needed
# pip install pulp

from pulp import *
import matplotlib.pyplot as plt
```

```python
# create optimization model

model = LpProblem(
    "Factory_Profit_Maximization",
    LpMaximize
)
```

```python
# how many units to produce

A = LpVariable(
    "Product_A",
    lowBound=0,
    cat="Integer"
)

B = LpVariable(
    "Product_B",
    lowBound=0,
    cat="Integer"
)
```

```python
# profits:
# A -> ₹300
# B -> ₹500

model += 300 * A + 500 * B
```

```python
# labor hours limit
model += 2 * A + 4 * B <= 100

# raw material limit
model += 3 * A + 2 * B <= 90
```

```python
model.solve()
```

```python
print("Status:", LpStatus[model.status])

print("Optimal Product A Units:", A.varValue)

print("Optimal Product B Units:", B.varValue)

print("Maximum Profit: ₹", value(model.objective))
```

```python
products = ["Product A", "Product B"]

values = [A.varValue, B.varValue]

plt.figure(figsize=(6,4))

plt.bar(products, values)

plt.title("Optimal Product Quantities")

plt.ylabel("Units")

plt.show()
```

# Business Insights

- Product B generates higher profit per unit.
- Resource constraints limit total production capacity.
- The optimization model helps determine the most profitable production mix.
- Linear programming can improve operational efficiency and decision-making.
